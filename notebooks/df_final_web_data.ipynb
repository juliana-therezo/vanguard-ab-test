```python
import pandas as pd
```

```python
path_pt1 = "C:\\Users\\Raquel Marques\\Desktop\\Ironhack\\Second Project\\df_final_web_data_pt_1.txt"

path_pt2 = "C:\\Users\\Raquel Marques\\Desktop\\Ironhack\\Second Project\\df_final_web_data_pt_2.txt"
```

```python
df_web1 = pd.read_csv(path_pt1, sep=",")

df_web2 = pd.read_csv(path_pt2, sep=",")
```

##### Merge both parts into one DataFrame

```python
df_web = pd.concat([df_web1, df_web2], ignore_index=True)

print("Dataset 1:", df_web1.shape)
print("Dataset 2:", df_web2.shape)
print("Merged dataset:", df_web.shape)
```

#### Duplicate Rows Analysis

- Display some duplicated rows

```python
duplicates = df_web[df_web.duplicated(keep=False)]

display(duplicates.sort_values(by=["client_id", "date_time"]).head(10))
```

- Percentage of duplicated rows

```python
duplicates_percentage = df_web.duplicated().sum() / len(df_web) * 100

print(f"Percentage of duplicates: {duplicates_percentage:.2f}%")
```

- Creation of a cleaned copy without exact duplicate rows

```python
df_web_clean = df_web.drop_duplicates().copy()

print("Original shape:", df_web.shape)
print("Cleaned shape:", df_web_clean.shape)
```

#### Convertion date_time to datetime

```python
df_web_clean["date_time"] = pd.to_datetime(df_web_clean["date_time"],errors="coerce")

print(df_web_clean["date_time"].dtype)
```
